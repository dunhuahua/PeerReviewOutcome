```python
import pandas as pd

try:
    df1 = pd.read_csv('combined.csv') # combined sheet 5 and 6 of 2017 document
    df2 = pd.read_csv('categorized_papers.csv') # paper with category and section content.

    # Merge the two DataFrames based on the specified columns
    merged_df = pd.merge(df1, df2, left_on='paper_id', right_on='Paper ID', how='inner')

    # Save the merged DataFrame to a new CSV file
    merged_df.to_csv('merged_papers.csv', index=False)

    print("Successfully merged 'combined.csv' and 'categorized_papers.csv' based on 'paper_id' and 'Paper ID' into 'merged_papers.csv'")

except FileNotFoundError as e:
    print(f"Error: One or both of the specified files ('combined.csv', 'categorized_papers.csv') were not found.")
except KeyError as e:
    print(f"Error: One or both of the specified columns ('paper_id', 'Paper ID') were not found in the DataFrames.")
except Exception as e:
    print(f"An error occurred: {e}")
```
